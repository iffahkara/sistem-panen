```python
def total_panen(jumlah, harga):
    return jumlah * harga

total = total_panen(10, 5000)
print("Total hasil panen: Rp", total)
```

```python
def hitung_diskon(total, diskon):
    return total - (total * diskon / 100)

print("Total setelah diskon: Rp", hitung_diskon(total, 10))
```

```python
def input_data():
    nama = input("Masukkan nama hasil panen: ")
    jumlah = float(input("Masukkan jumlah panen (kg): "))
    harga = float(input("Masukkan harga per kg: "))

    return nama, jumlah, harga

data = input_data()
print("Data panen:", data)
```
